Dicionários Exemplos Teoria


```python
# Acessando dados
contato = {'@luciano_ws': 'Luciano', '@joaozinho123': 'João Paulo'}

print(contato['@luciano_ws'])
```

```python
# Outra Maneira de acessar dados
contato = {}
contato['@luciano_ws'] = 'Luciano'

print(contato['@luciano_ws'])
```

```python
# Outra maneira é usando o get()
contato = {'luciano123': 'luciano',
'paolaoliveira4': 'Paola Oliveira'}

print(contato.get('paolaoliveira4'))
```

```python
# Adicionando elementos update(), muda valores e tambem adiciona novos valores
menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2'}

menu.update({'kibe' : 'alimento3'})
menu.update({'pao de queijo': 'alimento3'})

print(menu['pao de queijo'])
print(menu)
```

```python
# Contando Elementos

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2'}

print(len(menu))
```

```python
# Verificando a existencia de chaves

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2'}

if 'pao de queijo' in menu:
  print(f'{menu.get('pao de queijo')}')
```

```python
# Obtendo Valores e chaves - values() e keys()

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2', 'kibe' : 'alimento3', 'empada' : 'alimento4'}

for lista in menu.keys():
  print(lista)

# menu = {'pao de queijo' : 'alimento1',
# 'esfirra' : 'alimento2',
# 'kibe' : 'alimento3',
# 'empada' : 'alimento4'}

# for valores in menu.values():
#   print(valores)
```

```python
# Usando o .items() retorna a chave e o valor

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2',
'kibe' : 'alimento3',
'empada' : 'alimento4'}

for lista, valores in menu.items():
  print(f'{lista:<13} : {valores:>10}')
```

```python
# Ordenando Elementos pelas chaves

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2',
'kibe' : 'alimento3',
'empada' : 'alimento4'}

for lista,valores in sorted(menu.items()):
  print(f'{lista:<13} : {valores:>10}')
```

```python
# Ordenando Elementos pelos valores
from operator import itemgetter

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2',
'kibe' : 'alimento3',
'empada' : 'alimento4'}

for lista,valores in sorted(menu.items(), key = itemgetter(1)): # O valor 1 é para ordenar pelos valores, já o 0 é para ordenar conforme as chaves, porem o sorted faz isso
  print(f'{lista:<13} : {valores:>10}')
```

```python
# Clonando Dicionarios (dict)

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2',
'kibe' : 'alimento3',
'empada' : 'alimento4'}

copia = dict(menu)
print(copia)
```

```python
# Removendo Itens - .pop()

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2',
'kibe' : 'alimento3',
'empada' : 'alimento4'}

menu.pop('empada')
print(menu)
```

```python
# Esvaziando Dicionarios

menu = {'pao de queijo' : 'alimento1',
'esfirra' : 'alimento2',
'kibe' : 'alimento3',
'empada' : 'alimento4'}

copia = dict(menu) # Antes de esvaziar é interessante fazer uma copia, pois todos itens sao apagados
menu.clear()

print(menu)
print(copia)
```

EXERCICIO 01 - Construa um programa que utilize um dicionário cujas chaves são
os códigos do produto e os valores são o nome do produto, o preço
unitário e a quantidade comprada, como no exemplo a seguir.

Captura de tela 2026-06-08 125917.png

A partir do dicionário, o programa deve imprimir os itens da compra e calcular o subtotal de cada um deles, ou seja, quantidade * preço unitário. Por fim, o programa deve apresentar o valor total da compra

```python
from operator import itemgetter

codigos = {
      '1' : ['Monitor led 24', 599.99, 1],
      '2' : ['Teclado Wireless', 49.26, 1],
      '3' : ['Mouse Wireless', 19.90, 1],
      '4' : ['Cartucho Colorido', 54.00, 2]
}

# print(f'{'Código':<10} {'Nome':<20} {'Preço Unitário':>15} {'Qtde Comprada':>15}')
# Essa linha é para imprimir como se fosse a tabela
total = 0

for codigo,especificacoes in sorted(codigos.items(), key = itemgetter(0)):
  # print(f'{codigo:<10} {especificacoes[0]:<20} {especificacoes[1]:>15.2f} {especificacoes[2]:>15.2f}')
  # Essa linha é o complemento da outra
  subtotal = especificacoes[1] * especificacoes[2]
  total += subtotal
  print(f'{especificacoes[0]} : R$ {subtotal:.2f}')
  # Essa linha responde o que o exercicio quer, apenas o subtotal de cada item ( Quantos foram comprados e o preço deles gera o subtotal)

print()

print(f'O total das compras foi: {total:.2f}')

```

EXERCICIO 02 -No exemplo anterior, criamos um dicionário contendo quatro elementos que foram definidos por nós mesmos. Porém, na prática, ocorre que os dados são informados pelo usuário. Portanto, neste exercício, modificaremos o programa anterior de modo que os dados sejam informados pelo usuário.

```python
from operator import itemgetter
codigos = {}
menu = '''
Opção (A): Adicionar um item
Opção (B): Remover um item
Opção (C): Mostrar tabela
Opção (D): Mostrar Subtotal
Opção (E): Mostrar Valor Total
Opção (S): Sair
'''

while True:
  opcao = input('Digite uma opção - (m) para exibir o menu: ').strip().lower()
  print()

  if opcao == 's':
    print('Programa Encerrado!')
    print()
    break

  elif opcao == 'm':
    print(menu)
    print()

  elif opcao == 'a':
    try:
      cod = int(input('Código: '))
      nome = input('Nome: ')
      preco = float(input('Preço Unitário: '))
      qtde = float(input('Quantidade: '))

      # produto = []
      # produto.append(nome)
      # produto.append(preco)
      # produto.append(qtde)
      # codigos.update({cod : produto})
      # ESTA CERTO POREM TEM UMA MANEIRA MAIS LIMPA DE FAZER QUE EU NAO SABIA

      codigos[cod] = [nome, preco, qtde]

      print('Produto adicionado com sucesso!')
      print()

    except ValueError:
      print('O codigo, o preço e a quantidade devem ser numeros!')
      print()

  elif opcao == 'b':

    if codigos:
      try:
        removido = int(input('Digite o codigo do item que deseja remover: '))
        print()
        if removido in codigos:
          codigos.pop(removido)
          print('Item removido com sucesso!')
          print()
        else:
          print('Não faz parte da lista!')
          print()
      except ValueError:
        print('Digite um valor numerico para o codigo!!')
        print()

    else:
      print('Lista vazia!')
      print()

  elif opcao == 'c':

    if codigos:
      print(f'{"Código":<10} {"Nome":<20} {"Preço Unitário":>15} {"Qtde Comprada":>15}')
      for codigo, especificacoes in sorted(codigos.items(), key = itemgetter(0)):
        print(f'{codigo:<10} {especificacoes[0]:<20} {especificacoes[1]:>15.2f} {especificacoes[2]:>15.2f}')
      print()
    else:
      print('Lista Vazia!!')
      print()

  elif opcao == 'd':

    if codigos:
      for codigo, especificacoes in sorted(codigos.items(), key = itemgetter(0)):
        subtotal = especificacoes[1] * especificacoes[2]
        print(f'{especificacoes[0]} : R$ {subtotal:.2f}')
        print()
    else:
      print('Lista Vazia!!')
      print()

  elif opcao == 'e':

    if codigos:
      total_geral = sum(esp[1] * esp[2] for esp in codigos.values())
      print(f'O total das compras foi: R$ {total_geral:.2f}')
      print()
    else:
      print('Lista Vazia!!')
      print()

  else:
    print('Opção Inválida!!')
    print()
```

EXERCICIO 03 - Crie um programa para cadastrar matrícula, nome e salário dos funcionários de uma empresa. Ao fim, o programa deve mostrar todos os funcionários cadastrados e, em seguida, aumentar para 1.800,00 o salário dos funcionários com salários inferiores a 1.500,00. Depois do aumento salarial, o programa deve exibir a nova listagem.

```python
from operator import itemgetter
funcionarios = {}

while True:
  opcao = input('[S] -  Cadastrar Funcionario [N] - Encerrar programa: ').lower().strip()
  print()

  if opcao == 's':

    try:
      print('Dados Funcionário')
      matricula = int(input('Matrícula: '))
      nome = input('Nome: ')
      salario = float(input('Salario: '))
      print()

      funcionarios[matricula] = [nome, salario] #Cadastra no dicionário

      # dados = []
      # dados.append(nome)
      # dados.append(salario)

      # funcionarios.update({matricula: dados})
      # Maneira do livro

    except ValueError:
      print('Matrícula e Salarios devem ser números!!')
      print()

  elif opcao == 'n':
    if not funcionarios:
      print('Não há funcionários cadastrados!')
      break

    print('*'*15, 'Relação Empregados', '*'*15)
    print(f'{"Matrícula":<12} {"Nome":<20} {"Salário":>12}')
    for chave, valor in sorted(funcionarios.items(), key = itemgetter(0)):
      print(f'| {chave:<12} | {valor[0]:<20} | {valor[1]:>9.2f} R$ |')
    print()

    for chave,valor in funcionarios.items():
      if valor[1] < 1500:
        valor[1] = 1800

    print('*'*15, 'SALÁRIOS ATUALIZADOS', '*'*15)
    print(f'{"Matrícula":<12} {"Nome":<18} {"Salário":>12}')
    for chave, valor in sorted(funcionarios.items(), key = itemgetter(0)):
       print(f'| {chave:<12} | {valor[0]:<20} | {valor[1]:>9.2f} R$ |')
    print()
    break
```

Exercícios Propostos

Um usuário do departamento de vendas de uma empresa necessita de um relatório que apresente seus clientes potenciais. Para isso, é necessário que o relatório seja ordenado do cliente que mais comprou para o que menos comprou. Os dados de
entrada são razão social e valor total de compras. Considere a razão social como sendo a chave identificadora do cliente.

```python
from operator import itemgetter
clientes = {}

while True:
  opcao = input('Inserir novos dados[S] | Mostrar Tabela e encerrar[N]: ').strip().lower()

  if opcao == 's':
    razao = input('Nome da Empresa: ')
    vendas = float(input('Quantidade de Compras: '))

    clientes.update({razao : vendas})

  elif opcao == 'n':
    print(f'{"Nome Empresa":<20} | {"Compras":>12}')
    for chave,valor in sorted(clientes.items(), key = itemgetter(1), reverse = True):
      print(f'{chave:<20} | {valor:>12.2f}')
    break
```
