Crie um programa para gerar duas matrizes quadradas aleatórias (A e B), no intervalo [1, 10], de mesma ordem, a qual deve ser informada pelo usuário. Ao fim, o programa deve calcular e imprimir a soma entre os elementos de A que estão abaixo da diagonal principal com os elementos de B que estão acima da diagonal principal.


```python
from random import randrange
import random

linhas = int(input('Escreva o número de linhas: '))
colunas = int(input('Escreva o número de colunas: '))

if linhas == colunas:
  A = [[randrange(1,11) for i in range (linhas)] for j in range (colunas)]
  B = [[randrange(1,11) for i in range (linhas)] for j in range (colunas)]

  soma_abaixo = 0
  soma_acima = 0

  for i in range (linhas):
    for j in range (colunas):
      if i > j:
        soma_abaixo += A[i][j]
      if i < j:
        soma_acima += B[i][j]

  print(f'Matriz A:')
  for i in range (linhas):
    for j in range (colunas):
      print(A[i][j], end ='')
    print()


  print(f'Matriz B:')
  for i in range (linhas):
    for j in range (colunas):
      print(B[i][j], end = '')
    print()

  print(f'Soma = {soma_abaixo + soma_acima}')
else:
  print('Somente Matrizes quadradas!')
```

Questão 02 - Crie um programa no qual o usuário digite o número de
linhas e o número de colunas de uma matriz M. Ao fim, o programa deve informar se M é uma matriz triangular inferior.

```python
#Matriz triangular inferior - é quando os todos elementos ACIMA da diagonal principal(Matriz Quadrada) são iguais a 0.

from random import randrange
import random

linhas = int(input('Digite o número de linhas: '))
colunas = int(input('Digite o número de colunas: '))

print()

if linhas == colunas:
  m = []
  triangular = True

  for i in range (linhas):
    elem = []
    for j in range (colunas):
      elem.append(int(input(f'M[{i+1}] [{j+1}]:')))
    m.append(elem)

  print()

  print('Matriz M: ')
  for i in range(linhas):
    for j in range(colunas):
      print(m[i][j], end = '')
    print()

  for i in range (linhas):
    for j in range (colunas):
      if i < j:
        if m[i][j] != 0:
          triangular = False
          break

  if triangular == True:
    print('Matriz triangular inferior')
  else:
    print('Matriz não triangular inferior')

else:
  print('Matriz não é triangular!')



```

Um professor de Matemática deseja construir um programa para gerar uma Progressão Aritmética(PA). Para isso, devem ser informados 3 parâmetros deentrada:
a) primeiro termo da PA, b) quantidade de termos da PA e c) razão dessa PA.
Construa um programa para carregar e imprimir uma lista contendo os termos da PA,bem como a soma dos elementos da PA.

```python
a1 = int(input('Primeiro termo da P.A: '))
n = int(input('Quantidade de termos da P.A: '))
r = int(input('Razão da P.A: '))

pa = [(a1 + (r*(i-1))) for i in range(1, n+1)]
print(pa)

soma = sum(pa)
print(soma)
```

Os professores de Educação Física estão organizando uma
seletiva para montar a equipe de natação. Para isso, eles convocaram os 7 melhores tempos da última competição e marcaram o tempo de cada um dos nadadores, na prova dos 25 metros, estilo nado livre.

Considerando que não houve tempos iguais, construa um programa que leia o nome e o tempo (em segundos) de cada atleta e, em seguida, gere o seguinte relatório:

a.  nadador com o melhor tempo;

b.  nadador com o pior desempenho;

c.  tempo médio dos nadadores e;

Dica: Crie duas listas: uma para cadastrar os nomes dos nadadores e outra para guardar os seus respectivos tempos.

```python
import random

nomes = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
tempos = [random.randint(10,30) for x in range(len(nomes))]


lista_competicao = sorted(zip(tempos,nomes))

print('Tabela------')

for tempo,nome in lista_competicao:
  print(f'{nome} - {tempo}')

print()

tempo_melhor, nome_melhor = lista_competicao[0]
print(f'O nadador {nome_melhor} com o melhor desempenho fez os 25 metros em {tempo_melhor} segundos')

print()

tempo_pior, nome_pior = lista_competicao[-1]
print(f'O nadador {nome_pior} com o pior desempenho fez os 25 metros em {tempo_pior} segundos')

print()

media = sum(tempos) / len(tempos)
print(f'A média dos tempos dos nadadores foi: {media:.2f}')
```

Ao final, calcule e escreva o fatorial do elemento selecionado.

```python
import math
numero_impar = int(input('Digite um numero ímpar maior que 1: '))
lista = [int(input('Digite numeros inteiros e positivos: ')) for i in range (numero_impar)]

indice_numero_meio = numero_impar//2
numero_meio = lista[indice_numero_meio]

print(f'O numero do meio é: {numero_meio}')

numero_fatorado = [x for x in range(1, numero_meio+1)]
fatorial = math.prod(numero_fatorado)

print(f'O fatorial de {numero_meio} é {fatorial}')
```

Crie um programa que gere, aleatoriamente, uma matriz M, com elementos no intervalo [0,1]. A quantidadede linhas e de colunas de M devem ser informadas pelo usuário. Ao fim, o programa deve informar se M é uma matriz nula.

```python
from random import randrange

linhas = int(input('Digite a quantidade de linhas: '))
colunas = int(input('Digite a quantidade de colunas: '))

print()
#OBS: O LOOP DE DENTRO DAS CHAVES É TAMANHO DAS SUBLISTAS(QUANTOS ITENS TERÃO DENTRO DELAS(OU SEJA AS !COLUNAS!)) JA AS DE FORA SERAO AS LINHAS(QUANTAS SUBLISTAS TERAO)
M = [[randrange(0,2) for i in range(colunas)]for j in range(linhas)]

matriznula = True #ASSUMIR QUE UMA COISA É VERDADEIRA ATE QUE SE PROVE O CONTRARIO É MAIS EFICIENTE NESSE CASO

print('Matriz M')
for linha in range(linhas):
  for coluna in range(colunas):
    print(M[linha][coluna], end = ' ')
  print()

print()

for elementos in M:
  for i in elementos:
    if i != 0:
      matriznula = False
      break

if matriznula:
  print('Matriz Nula')
else:
  print('Matriz Não Nula')

```

Crie um programa no qual o usuário informe o número de linhas e o número de colunas de uma matriz M e, em seguida, o usuário deve digitar os elementos de M. Ao fim, o programa deve informar se M é uma matriz identidade.

```python
linhas = int(input('Digite o numero de linhas: '))
colunas = int(input('Digite o numero de colunas: '))

if linhas == colunas:
  m = []
  identidade = True

  for i in range(linhas):
    elemento = [] # Esvazia a lista elemento a cada loop
    for j in range(colunas):
      elemento.append(int(input(f'M[{i+1}] [{j+1}]:'))) # Coloca a quantidade de elementos que cada lista terá
    m.append(elemento)

  print()

  print('Matriz M:')
  for i in range (linhas):
    for j in range (colunas):
      print(m[i][j], end = ' ')
    print()

  for i in range (linhas):
    for j in range (colunas):
      if i == j and m[i][j] != 1:
          identidade = False
      elif i != j and m[i][j] != 0:
          identidade = False

  if identidade:
    print('Matriz identidade')
  else:
    print('Matriz Normal')

else:
  print('Matriz deve ser quadrada!')

```

Crie um programa que gere, aleatoriamente, uma matriz M. A quantidade de linhas e de colunas de M devem ser informadas pelo usuário. Ao fim, o programa deve informar se M é uma matriz diagonal.

```python
from random import randrange

linhas = int(input('Digite o numero de linhas: '))
colunas = int(input('Digite o numero de colunas: '))

if linhas == colunas:
  M = [[randrange(0, 10) for i in range (colunas)] for j in range(linhas)]
  matrizdiagonal = True

  print()

  for i in range(linhas):
    for j in range(colunas):
      if i != j and M[i][j] != 0:
        matrizdiagonal = False

  for i in range (linhas):
    for j in range (colunas):
      print(M[i][j], end = ' ')
    print()

  print()

  if matrizdiagonal:
    print('Matriz diagonal!')
  else:
    print('Matriz Normal')

else:
  print('Matriz deve ser quadrada!')
```

Uma matriz é considerada esparsa quando a quantidade de elementos iguais a zero supera a quantidade de elementos diferentes de zero. Crie um programa que gere, aleatoriamente, uma matriz M. A quantidade de linhas e de colunas de M devem ser informadas pelo usuário. Ao fim, o programa deve informar se M é uma matriz esparsa.

```python
from random import randrange

linhas = int(input('Linha: '))
colunas = int(input('Colunas: '))
esparsa = False
zero = 0
numero = 0

M = [[randrange(0,2) for i in range(colunas)]for j in range(linhas)]

print()

print('Matriz M:')
for i in range(linhas):
  for j in range(colunas):
    print(M[i][j], end = ' ')
  print()

for i in range(linhas):
  for j in range(colunas):
    if M[i][j] == 0:
      zero += 1
    else:
      numero += 1

if zero > numero:
  esparsa = True

if esparsa:
  print('Matriz Esparsa')
else:
  print('Matriz Normal')
```
